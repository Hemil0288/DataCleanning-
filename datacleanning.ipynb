```python
# important step
# identify categorical columns in a dataset and display the frequency of each unique value.
# nunique() counts how many different values are present in a column.
# < 20 means less than 20.
# Example
# Suppose we have a Gender column:
# Gender
# Male
# Female
# Male
# Female
# Male
for col in df.columns:
  if df[col].nunique()<20:
    print(col)
    print(df[col].value_counts())



```

```python
#1] handling missing value

#1] handling missing value in customer data
# drop the missing(null) value in specific customer_id
df.dropna(subset=["Customer_ID"],inplace=True)
```

```python
df.isna().sum()
```

```python
#2]handling the customer age
df["Age"].unique()
```

```python
import re
def extract_age(age):
  age_num=re.findall('[0-9]+',str(age))
  if len(age_num)>0:
    return age_num[0]
  else:
    return age
df["Age"]=df["Age"].apply(lambda x:extract_age(x))
```

```python
df["Age"].unique()
```

```python
#fill the nan years vakue into take of all median
df_age=df[df["Age"]!="nan years"]["Age"]
```

```python

age_median=int(df_age.dropna().astype('int64').median())
```

```python
age_median
```

```python
# now repalce the nan years to age_median
df.replace("nan years",age_median,inplace=True)
```

```python
df["Age"].unique()
```

```python
# to fill other nan value to median
df["Age"].fillna(age_median,inplace=True)
```

```python
df["Age"].unique()
```

```python
# 2]handle the purchase amount
df["Purchase_Amount"].unique()

# fill the not null or empty value in median directly because it is a integr
purchae_median=df["Purchase_Amount"].median()
df["Purchase_Amount"].fillna(purchae_median,inplace=True)
```

```python
df.isna().sum()
```

```python
# 3]handle the feedback
df["Feedback_Score"]
```

```python
df.info()
```

```python
# here in feeb back score to  repalce the null value into the mode because of its represent as categorical discreate  value

feedbck_mode=df["Feedback_Score"].mode()[0]
df["Feedback_Score"].fillna(feedbck_mode,inplace=True)
```

```python
df.isna().sum()
```

```python
# 4]now handle the all missing value in gender & city & country
for col in ["Gender","City","Country"]:
  df[col].fillna(df[col].mode()[0],inplace=True)
```

```python
df.isna().sum()
```

```python
# 5]handle the last_purchase_date using forwadl fill
df["Last_Purchase_Date"].fillna(df["Last_Purchase_Date"].ffill(),inplace=True)
```
